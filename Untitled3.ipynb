```python
import os
from collections import defaultdict
from PIL import Image
import torchvision.transforms as transforms
import random

# ────────────────────────────────────────────────
# CONFIG - CHANGE THIS TO YOUR FOLDER PATH
# ────────────────────────────────────────────────
FOLDER = r"C:\Users\gupta\7th SEM\Project\Signature\content\testfin\train_sig\full_org"  # ← edit this
TARGET_COUNT = 10  # increase to this many per aid
DRY_RUN = False     # Set False to actually save new images

# Augmentation pipeline (adjust as needed)
augment_transform = transforms.Compose([
    transforms.Pad(6, fill=255, padding_mode='constant'),
    transforms.RandomAffine(degrees=15, translate=(0.05, 0.05),fill=255),
    transforms.RandomPerspective(distortion_scale=0.1, p=1, fill=255),
    transforms.Resize((96, 96)),
    transforms.ToTensor(),
    transforms.ToPILImage(),  # back to PIL for saving
])

# ────────────────────────────────────────────────

def main():
    # Group files by aid
    files_by_aid = defaultdict(list)
    max_n_by_aid = defaultdict(int)

    for filename in os.listdir(FOLDER):
        if not filename.lower().endswith('.png'):
            continue

        parts = filename.split('_')
        if len(parts) != 3 or parts[0] != 'genuine':
            continue

        aid_str = parts[1]
        n_str = parts[2].rsplit('.', 1)[0]  # remove .png

        if not aid_str.isdigit() or not n_str.isdigit():
            continue

        files_by_aid[aid_str].append(filename)
        max_n_by_aid[aid_str] = max(max_n_by_aid[aid_str], int(n_str))

    print(f"Found {len(files_by_aid)} unique aids\n")

    created = 0
    skipped_aids = 0

    for aid_str, files in sorted(files_by_aid.items()):
        current_count = len(files)
        print(f"Aid {aid_str}: {current_count} images (max N={max_n_by_aid[aid_str]})")

        if current_count >= TARGET_COUNT:
            skipped_aids += 1
            continue

        to_create = TARGET_COUNT - current_count
        print(f"  Creating {to_create} augmented images...")

        # Randomly select existing images to augment
        source_files = random.choices(files, k=to_create)

        next_n = max_n_by_aid[aid_str] + 1

        for source_filename in source_files:
            new_filename = f"genuine_{aid_str}_{next_n:03d}.png"  # adjust padding :03d as needed

            old_path = os.path.join(FOLDER, source_filename)
            new_path = os.path.join(FOLDER, new_filename)

            if os.path.exists(new_path):
                print(f"    Conflict: {new_filename} exists → skipping")
                continue

            img = Image.open(old_path).convert('RGB')  # or 'L' for grayscale

            # Apply augmentation
            augmented_img = augment_transform(img)

            action = "Would save" if DRY_RUN else "Saved"
            print(f"    {action}: {new_filename} (from {source_filename})")

            if not DRY_RUN:
                augmented_img.save(new_path)
                created += 1

            next_n += 1

    print("\n" + "═"*60)
    print(f"Finished.")
    print(f"  Aids processed: {len(files_by_aid) - skipped_aids}")
    print(f"  New images created: {created}")
    print(f"  Aids skipped (already >=10): {skipped_aids}")
    if DRY_RUN:
        print("DRY RUN ONLY — no new files were created. Set DRY_RUN = False to save.")

if __name__ == "__main__":
    if not os.path.isdir(FOLDER):
        print(f"Error: Folder not found → {FOLDER}")
    else:
        main()
```
