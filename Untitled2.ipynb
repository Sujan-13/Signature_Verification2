```python
import os
import cv2
import numpy as np

folders = [r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_org", r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_forg"]

# Define target lengths with more tolerance
target_horizontal_length = 205
target_vertical_length = 140
tolerance = 20  # Increased to +/- 20 pixels

for folder in folders:
    for filename in os.listdir(folder):
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
            continue

        path = os.path.join(folder, filename)

        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        h, w = img.shape
        cleaned = img.copy()

        # Pre-process: Gaussian blur to reduce noise
        blurred = cv2.GaussianBlur(img, (5, 5), 0)

        # Detect edges with adjusted Canny parameters
        edges = cv2.Canny(blurred, 30, 100, apertureSize=3)

        # Detect straight lines with lowered threshold and minLineLength
        lines = cv2.HoughLinesP(
            edges,
            rho=1,
            theta=np.pi/180,
            threshold=20,  # Lowered threshold
            minLineLength=100,  # Fixed lower min length
            maxLineGap=0  # Increased gap to merge broken lines
        )

        removed_count = 0
        if lines is not None:
            for line in lines:
                x1, y1, x2, y2 = line[0]

                # Check if line is approximately horizontal
                if abs(y1 - y2) < 30:  # Allow more slant
                    if y1 < 20 or y1 > h - 20:
                        length = abs(x1 - x2)
                        if abs(length - target_horizontal_length) <= tolerance:
                            cv2.line(cleaned, (x1, y1), (x2, y2), 255, 10)  # Thicker erase
                            removed_count += 1

                # Check if line is approximately vertical
                if abs(x1 - x2) < 30:  # Allow more slant
                     if x1 < 50 or x1 > w - 50:
                        length = abs(y1 - y2)
                        if abs(length - target_vertical_length) <= tolerance:
                            cv2.line(cleaned, (x1, y1), (x2, y2), 255, 10)
                            removed_count += 1

        cv2.imwrite(path, cleaned)
        print(f"Cleaned: {filename} (removed {removed_count} lines)")

print("✅ Specific line removal complete with adjusted parameters.")
```

```python
import os
import cv2
import numpy as np

folders = [r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_org", r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_forg"]

edge_margin = 50  # Pixels from edge to consider as border

for folder in folders:
    for filename in os.listdir(folder):
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
            continue

        path = os.path.join(folder, filename)

        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        h, w = img.shape
        cleaned = img.copy()

        # Pre-process: Gaussian blur to reduce noise
        blurred = cv2.GaussianBlur(img, (5, 5), 0)

        # Detect edges with adjusted Canny parameters
        edges = cv2.Canny(blurred, 30, 100, apertureSize=3)  # Lower thresholds to detect faint lines

        # Detect straight lines with adjusted parameters
        lines = cv2.HoughLinesP(
            edges,
            rho=1,
            theta=np.pi/180,
            threshold=20,  # Lower threshold to detect more lines
            minLineLength=100,  # Minimum length to consider
            maxLineGap=20  # Gap to merge broken lines
        )

        removed_count = 0
        if lines is not None:
            for line in lines:
                x1, y1, x2, y2 = line[0]

                # Calculate length
                length = np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

                if length < 100:
                    continue

                # Check if approximately horizontal
                if abs(y1 - y2) < 20:
                      if y1 < 50 or y1 > h - 50: 
                        avg_y = (y1 + y2) // 2
                        if avg_y < edge_margin or avg_y > h - edge_margin:
                            cv2.line(cleaned, (x1, y1), (x2, y2), 255, 10)  # Erase with thicker line
                            removed_count += 1

                # Check if approximately vertical
                if abs(x1 - x2) < 20:
                    if x1 < 50 or x1 > w - 50:
                        avg_x = (x1 + x2) // 2
                        if avg_x < edge_margin or avg_x > w - edge_margin:
                            cv2.line(cleaned, (x1, y1), (x2, y2), 255, 10)
                            removed_count += 1

        cv2.imwrite(path, cleaned)
        print(f"Cleaned: {filename} (removed {removed_count} lines)")

print("✅ Edge line removal complete with position-based filtering.")
```

```python
import cv2
import os
import numpy as np

folders = [r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_org", r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_forg"]

BORDER_WIDTH = 25  # width in pixels to check at edges

for folder in folders:
    for filename in os.listdir(folder):
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
            continue

        path = os.path.join(folder, filename)
        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        h, w = img.shape
        cleaned = img.copy()

        # Remove top border
        top = cleaned[0:BORDER_WIDTH, :]
        if np.mean(top) < 250:  # mostly black line
            cleaned[0:BORDER_WIDTH, :] = 255

        # Remove bottom border
        bottom = cleaned[h-BORDER_WIDTH:h, :]
        if np.mean(bottom) < 250:
            cleaned[h-BORDER_WIDTH:h, :] = 255

        # Remove left border
        left = cleaned[:, 0:BORDER_WIDTH]
        if np.mean(left) < 250:
            cleaned[:, 0:BORDER_WIDTH] = 255

        # Remove right border
        right = cleaned[:, w-BORDER_WIDTH:w]
        if np.mean(right) < 250:
            cleaned[:, w-BORDER_WIDTH:w] = 255

        cv2.imwrite(path, cleaned)
        print(f"Cleaned borders: {filename}")

print("✅ Edge borders removed successfully.")
```

```python
import os
import cv2
import numpy as np

folders = [r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_org", r"C:\Users\gupta\7th SEM\Project\Signature\content\testnew\full_forg"]

dark_threshold = 200  # Pixels below this are considered part of the signature
min_margin = 5  # Minimum margin in pixels from signature to edge
min_dark_pixels = 100  # Safety check, though empties are already removed

for folder in folders:
    for filename in os.listdir(folder):
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
            continue

        path = os.path.join(folder, filename)

        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        h, w = img.shape

        # Find positions of dark pixels
        rows, cols = np.where(img < dark_threshold)
        if len(rows) < min_dark_pixels:
            print(f"Skipping: {filename} (too few dark pixels)")
            continue

        min_y = np.min(rows)
        max_y = np.max(rows)
        min_x = np.min(cols)
        max_x = np.max(cols)

        # Calculate margins
        left = min_x
        top = min_y
        right = w - max_x - 1
        bottom = h - max_y - 1

        # Check if any margin is too small
        if min(left, top, right, bottom) < min_margin:
            os.remove(path)
            print(f"Deleted: {filename} (tight to edge: margins L{left}/T{top}/R{right}/B{bottom})")
        else:
            print(f"Kept: {filename} (margins L{left}/T{top}/R{right}/B{bottom})")

print("✅ Tight signature removal complete.")
```
