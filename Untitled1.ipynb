```python
"""
FIXED PREPROCESSING FOR SIGNATURE VERIFICATION
==============================================

The problem: Your current preprocessing results in images that are 96.7% white,
making signatures too faint for the model to learn from.

This script provides corrected preprocessing with:
1. Better cropping (removes excess white space)
2. Contrast enhancement (makes signatures darker)
3. Proper centering (signature fills more of the image)
"""

import numpy as np
from PIL import Image, ImageEnhance
import cv2
import os
import shutil


def improved_crop_and_save_signature(input_path, output_path, 
                                     target_size=64, 
                                     signature_size=64,
                                     enhance_contrast=1.3,
                                     apply_threshold=False):
    """
    Improved preprocessing that creates darker, more visible signatures
    
    Args:
        input_path: Path to input image
        output_path: Where to save processed image
        target_size: Final image size (default 64×64)
        signature_size: Size of signature within the image (default 56×56)
        enhance_contrast: Contrast multiplier (1.0=no change, 1.3=30% more contrast)
        apply_threshold: If True, apply Otsu binarization for maximum contrast
    """
    
    # Load image
    img = Image.open(input_path).convert('L')
    arr = np.array(img)
    
    # Step 1: Find signature using Otsu threshold
    _, binary = cv2.threshold(arr, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    
    # Find bounding box
    coords = np.where(binary > 10)
    
    if len(coords[0]) == 0:
        # No signature found - just resize
        img_resized = img.resize((target_size, target_size), Image.Resampling.LANCZOS)
        img_resized.save(output_path, "PNG")
        return
    
    y0, y1 = coords[0].min(), coords[0].max() + 1
    x0, x1 = coords[1].min(), coords[1].max() + 1
    
    # Step 2: Add padding (15% on each side)
    h, w = y1 - y0, x1 - x0
    pad_y = int(h * 0.15)
    pad_x = int(w * 0.15)
    
    y0 = max(0, y0 - pad_y)
    y1 = min(arr.shape[0], y1 + pad_y)
    x0 = max(0, x0 - pad_x)
    x1 = min(arr.shape[1], x1 + pad_x)
    
    # Step 3: Crop to signature region (from ORIGINAL, not binary!)
    cropped = Image.fromarray(arr).crop((x0, y0, x1, y1))
    
    # Step 4: Apply contrast enhancement
    if enhance_contrast != 1.0:
        enhancer = ImageEnhance.Contrast(cropped)
        cropped = enhancer.enhance(enhance_contrast)
    
    # Step 5: Optional binarization for maximum contrast
    if apply_threshold:
        cropped_arr = np.array(cropped)
        _, cropped_binary = cv2.threshold(
            cropped_arr, 0, 255, 
            cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
        )
        kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (2, 2))
        opened = cv2.morphologyEx(cropped_binary, cv2.MORPH_OPEN, kernel)
    
        dilated = cv2.dilate(opened, kernel, iterations=15)
        cropped_binary = 255 - dilated  # Invert back (white background)
        cropped = Image.fromarray(cropped_binary)
    
    # Step 6: Resize signature to signature_size (leaving room for padding)
    cropped = cropped.resize((signature_size, signature_size), Image.Resampling.LANCZOS)
    
    # Step 7: Create white background and center the signature
    # final = Image.new('L', (target_size, target_size), 255)
    # paste_pos = (target_size - signature_size) // 2
    # final.paste(cropped, (paste_pos, paste_pos))
    final = cropped
    # Step 8: Save
    final.save(output_path, "PNG", optimize=True)


def batch_reprocess_dataset(input_dir, output_dir, 
                            enhance_contrast=1.3, 
                            apply_threshold=False,
                            verify_samples=5):
    """
    Reprocess entire dataset with improved preprocessing
    
    Args:
        input_dir: Directory with original images
        output_dir: Where to save processed images
        enhance_contrast: Contrast enhancement factor
        apply_threshold: Whether to apply binarization
        verify_samples: Number of before/after comparisons to show
    """
    
    print(f"\n{'='*60}")
    print(f"BATCH REPROCESSING: {input_dir}")
    print(f"{'='*60}")
    print(f"Settings:")
    print(f"  Contrast enhancement: {enhance_contrast}x")
    print(f"  Apply threshold: {apply_threshold}")
    print(f"  Output: {output_dir}")
    
    # Create output directory
    if os.path.exists(output_dir):
        response = input(f"\n⚠️  {output_dir} exists. Delete and recreate? (y/n): ")
        if response.lower() == 'y':
            shutil.rmtree(output_dir)
            print(f"Deleted {output_dir}")
        else:
            print("Aborting.")
            return
    
    os.makedirs(output_dir, exist_ok=True)
    
    # Get all image files
    files = [f for f in os.listdir(input_dir) 
             if f.endswith(('.png', '.PNG', '.jpg', '.jpeg', '.bmp'))]
    
    print(f"\nFound {len(files)} images")
    print("Processing...")
    
    # Process all files
    for idx, filename in enumerate(files):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)
        
        improved_crop_and_save_signature(
            input_path, 
            output_path,
            enhance_contrast=enhance_contrast,
            apply_threshold=apply_threshold
        )
        
        if (idx + 1) % 100 == 0:
            print(f"  Processed {idx + 1}/{len(files)}...")
    
    print(f"✓ Completed! Processed {len(files)} images")
    
    # Verify samples
    if verify_samples > 0:
        print(f"\n{'='*60}")
        print(f"VERIFICATION: Showing {verify_samples} before/after samples")
        print(f"{'='*60}")
        
        import matplotlib.pyplot as plt
        from torchvision import transforms
        
        transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.5,), (0.5,))
        ])
        
        for i, filename in enumerate(files[:verify_samples]):
            # Load before and after
            before_path = os.path.join(input_dir, filename)
            after_path = os.path.join(output_dir, filename)
            
            before = Image.open(before_path).convert('L')
            after = Image.open(after_path).convert('L')
            
            # Apply transform (what model sees)
            before_t = transform(before.resize((64, 64)))
            after_t = transform(after)
            
            before_np = (before_t[0].numpy() * 0.5) + 0.5
            after_np = (after_t[0].numpy() * 0.5) + 0.5
            
            # Visualize
            fig, axes = plt.subplots(2, 3, figsize=(12, 8))
            fig.suptitle(f'Sample {i+1}: {filename}', fontsize=14)
            
            # Before
            axes[0, 0].imshow(before, cmap='gray')
            axes[0, 0].set_title('Before (Original)')
            axes[0, 0].axis('off')
            
            axes[0, 1].imshow(before_np, cmap='gray', vmin=0, vmax=1)
            axes[0, 1].set_title(f'Before (Model Sees)\nMean={before_np.mean():.3f}')
            axes[0, 1].axis('off')
            
            axes[0, 2].hist(before_np.flatten(), bins=50, alpha=0.7, color='blue')
            axes[0, 2].set_title(f'Before\nSignature: {(before_np<0.9).sum()/(64*64)*100:.1f}%')
            axes[0, 2].axvline(0.5, color='red', linestyle='--')
            
            # After
            axes[1, 0].imshow(after, cmap='gray')
            axes[1, 0].set_title('After (Processed)')
            axes[1, 0].axis('off')
            
            axes[1, 1].imshow(after_np, cmap='gray', vmin=0, vmax=1)
            axes[1, 1].set_title(f'After (Model Sees)\nMean={after_np.mean():.3f}')
            axes[1, 1].axis('off')
            
            axes[1, 2].hist(after_np.flatten(), bins=50, alpha=0.7, color='green')
            axes[1, 2].set_title(f'After\nSignature: {(after_np<0.9).sum()/(64*64)*100:.1f}%')
            axes[1, 2].axvline(0.5, color='red', linestyle='--')
            
            plt.tight_layout()
            plt.show()
            
            print(f"\nSample {i+1} stats:")
            print(f"  Before: Mean={before_np.mean():.3f}, Signature={( before_np<0.9).sum()/(64*64)*100:.1f}%")
            print(f"  After:  Mean={after_np.mean():.3f}, Signature={(after_np<0.9).sum()/(64*64)*100:.1f}%")
            print(f"  Improvement: {(after_np<0.9).sum()/(64*64) / max((before_np<0.9).sum()/(64*64), 0.01):.1f}x more visible")


def reprocess_all_datasets(base_dir='./content', 
                           enhance_contrast=1.3, 
                           apply_threshold=False):
    """
    Reprocess all signature datasets
    
    Args:
        base_dir: Base directory containing PNG folder
        enhance_contrast: Contrast enhancement (1.3 = good default)
        apply_threshold: Use Otsu threshold (False = keep grayscale, True = binary)
    """
    
    # Define all directories to process
    datasets = [
        ('Offline Genuine', 'Offline Genuine'),
        ('Offline Forgeries', 'Offline Forgeries'),
    ]
    
    # Add reference and questioned directories if they exist
    reference_base = os.path.join(base_dir, 'Reference(646)')
    questioned_base = os.path.join(base_dir, 'Questioned(1287)')
    
    if os.path.exists(reference_base):
        for author_id in os.listdir(reference_base):
            author_path = os.path.join(reference_base, author_id)
            if os.path.isdir(author_path):
                datasets.append(
                    (f'Reference(646)/{author_id}', f'Reference(646)/{author_id}')
                )
    
    if os.path.exists(questioned_base):
        for author_id in os.listdir(questioned_base):
            author_path = os.path.join(questioned_base, author_id)
            if os.path.isdir(author_path):
                datasets.append(
                    (f'Questioned(1287)/{author_id}', f'Questioned(1287)/{author_id}')
                )
    
    print(f"\n{'='*70}")
    print("REPROCESSING ALL DATASETS WITH IMPROVED PREPROCESSING")
    print(f"{'='*70}")
    print(f"\nSettings:")
    print(f"  Contrast enhancement: {enhance_contrast}x")
    print(f"  Apply Otsu threshold: {apply_threshold}")
    print(f"\nDatasets to process: {len(datasets)}")
    
    response = input("\nContinue? (y/n): ")
    if response.lower() != 'y':
        print("Aborted.")
        return
    
    # Process each dataset
    for input_subdir, output_subdir in datasets:
        input_dir = os.path.join(base_dir, input_subdir)
        output_dir = os.path.join(base_dir, 'PNG_IMPROVED', output_subdir)
        
        if not os.path.exists(input_dir):
            print(f"\n⚠️  Skipping {input_dir} (not found)")
            continue
        
        batch_reprocess_dataset(
            input_dir, 
            output_dir,
            enhance_contrast=enhance_contrast,
            apply_threshold=apply_threshold,
            verify_samples=0  # Don't show samples for batch mode
        )
    
    print(f"\n{'='*70}")
    print("✓ ALL DATASETS REPROCESSED!")
    print(f"{'='*70}")
    print(f"\nNew images saved to: {base_dir}/PNG_IMPROVED/")
    print("\nNext steps:")
    print("1. Update your dataset paths to use PNG_IMPROVED folder")
    print("2. Retrain your model from scratch")
    print("3. Expected improvement: 60% → 85%+ accuracy")


if __name__ == "__main__":
    print("""
    ╔════════════════════════════════════════════════════════════╗
    ║         IMPROVED PREPROCESSING FOR SIGNATURES              ║
    ║  Fixes the "faint signature" problem (Mean=0.967)         ║
    ╚════════════════════════════════════════════════════════════╝
    """)
    
    print("\n📋 USAGE OPTIONS:\n")
    
    print("1. Test on single directory first:")
    print("   batch_reprocess_dataset(")
    print("       './content/Offline Genuine',")
    print("       './content/TEST_OUTPUT',")
    print("       enhance_contrast=1.3,")
    print("       apply_threshold=False,")
    print("       verify_samples=5")
    print("   )")
    
    print("\n2. Reprocess all datasets:")
    print("   reprocess_all_datasets(")
    print("       base_dir='./content',")
    print("       enhance_contrast=1.3,    # 30% more contrast")
    print("       apply_threshold=False     # Keep grayscale (recommended)")
    print("   )")
    
    print("\n💡 RECOMMENDATIONS:")
    print("  • enhance_contrast=1.3 is a good default (30% more contrast)")
    print("  • apply_threshold=False keeps grayscale (better for model)")
    print("  • apply_threshold=True gives maximum contrast (binary black/white)")
    print("  • Test on ONE directory first, check results, then process all")
```

```python
"""
AGGRESSIVE PREPROCESSING FOR VERY FAINT SIGNATURES
==================================================

When original scans are already light gray, we need aggressive processing.
"""
import numpy as np
from PIL import Image, ImageEnhance
import cv2
import os
import matplotlib.pyplot as plt
from torchvision import transforms


def aggressive_preprocessing(
    input_path,
    output_path,
    method='mild_clahe_otsu',      # changed default
    target_size=64,
    debug=False
):
    """
    Improved aggressive preprocessing — aims to KEEP mean high (~0.93+)
    while still darkening faint signature strokes enough to be <0.6–0.7.
    
    New / better methods:
      - 'mild_clahe_otsu'      ← usually best balance
      - 'clahe_light_dilate'
      - 'sauvola'
      - 'niblack'
      - 'contrast_stretch_otsu'
      - 'guided_otsu'          ← very clean background (slow)
    """
    img = Image.open(input_path).convert('L')
    arr = np.array(img, dtype=np.float32)

    # ──────────────────────────────────────────────
    # A. Common pre-enhancement steps
    # ──────────────────────────────────────────────
    # Very light denoising if extremely noisy
    arr = cv2.GaussianBlur(arr, (3,3), 0.6)

    if method in ['mild_clahe_otsu', 'clahe_light_dilate']:
        clahe = cv2.createCLAHE(clipLimit=1.8, tileGridSize=(8,8))   # milder clip
        enhanced = clahe.apply(arr.astype(np.uint8))
        
    elif method == 'contrast_stretch_otsu':
        p2, p98 = np.percentile(arr, (2, 98))
        if p98 > p2:
            arr_stretched = np.clip((arr - p2) / (p98 - p2) * 255, 0, 255).astype(np.uint8)
        else:
            arr_stretched = arr.astype(np.uint8)
        enhanced = arr_stretched
        
    elif method in ['sauvola', 'niblack']:
        enhanced = arr.astype(np.uint8)
        
    elif method == 'guided_otsu':
        # Guided filter preserves edges, smooths background
        from cv2.ximgproc import guidedFilter
        guided = guidedFilter(arr.astype(np.uint8), arr.astype(np.uint8), radius=10, eps=0.02*255**2)
        enhanced = guided
    else:
        enhanced = arr.astype(np.uint8)

    # ──────────────────────────────────────────────
    # B. Thresholding strategies
    # ──────────────────────────────────────────────
    if 'otsu' in method:
        _, binary = cv2.threshold(enhanced, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        processed = 255 - binary  # white bg, dark signature

    elif method == 'sauvola':
        from skimage.filters import threshold_sauvola
        thresh = threshold_sauvola(enhanced, window_size=15, k=0.35)
        binary = (enhanced < thresh).astype(np.uint8) * 255
        processed = 255 - binary

    elif method == 'niblack':
        from skimage.filters import threshold_niblack
        thresh = threshold_niblack(enhanced, window_size=15, k=0.2)
        binary = (enhanced < thresh).astype(np.uint8) * 255
        processed = 255 - binary

    elif method == 'clahe_light_dilate':
        # After mild CLAHE → light threshold + small dilation on ink
        _, binary = cv2.threshold(enhanced, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        kernel = np.ones((2,2), np.uint8)
        binary = cv2.dilate(binary, kernel, iterations=1)   # grow faint strokes slightly
        processed = 255 - binary

    elif method == 'strong_clahe_otsu':
        clahe = cv2.createCLAHE(clipLimit=5.0, tileGridSize=(8,8))  # higher clip → stronger local contrast
        enhanced = clahe.apply(arr.astype(np.uint8))
        _, binary = cv2.threshold(enhanced, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        processed = 255 - binary

    elif method == 'niblack_tuned':
        from skimage.filters import threshold_niblack
        # Smaller window = more local sensitivity, higher k = more aggressive (darker ink & bg)
        thresh = threshold_niblack(enhanced, window_size=9, k=0.4)   # try k=0.3 to 0.5
        binary = (enhanced < thresh).astype(np.uint8) * 255
        processed = 255 - binary
        # Light opening to clean bg noise
        kernel = np.ones((2,2), np.uint8)
        processed = cv2.morphologyEx(processed, cv2.MORPH_OPEN, kernel)
    
    elif method == 'background_subtract_otsu':
        # Rolling ball / morphological background estimation
        kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (25,25))  # tune size
        bg = cv2.morphologyEx(enhanced, cv2.MORPH_TOPHAT, kernel)       # or use BlackHat
        subtracted = cv2.subtract(enhanced, bg)
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
        subtracted = clahe.apply(subtracted)
        _, binary = cv2.threshold(subtracted, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        processed = binary  # here ink is white, bg black → invert if needed
        processed = 255 - processed
    
    elif method == 'strong_clahe_niblack':
        clahe = cv2.createCLAHE(clipLimit=4.5, tileGridSize=(8,8))
        enhanced = clahe.apply(arr.astype(np.uint8))
        from skimage.filters import threshold_niblack
        thresh = threshold_niblack(enhanced, window_size=11, k=0.35)
        binary = (enhanced < thresh).astype(np.uint8) * 255
        processed = 255 - binary
    


    else:
        raise ValueError(f"Unknown method: {method}")

    # ──────────────────────────────────────────────
    # C. Post-processing: remove small noise specks
    # ──────────────────────────────────────────────
    # Opening = erode then dilate → removes small white spots in background
    kernel = np.ones((2,2), np.uint8)
    processed = cv2.morphologyEx(processed, cv2.MORPH_OPEN, kernel, iterations=1)

    # Optional debug
    if debug:
        cv2.imwrite(f"debug_{method}_binary.png", processed)

    # ──────────────────────────────────────────────
    # D. Crop + resize + center (same as before)
    # ──────────────────────────────────────────────
    binary_for_crop = cv2.threshold(processed, 200, 255, cv2.THRESH_BINARY_INV)[1]
    coords = np.where(binary_for_crop > 10)

    if len(coords[0]) == 0:
        final = Image.fromarray(processed).resize((target_size, target_size), Image.Resampling.LANCZOS)
        final.save(output_path, "PNG")
        return

    y0, y1 = coords[0].min(), coords[0].max() + 1
    x0, x1 = coords[1].min(), coords[1].max() + 1

    h, w = y1 - y0, x1 - x0
    pad_y = int(h * 0.18)
    pad_x = int(w * 0.18)

    y0, y1 = max(0, y0 - pad_y), min(processed.shape[0], y1 + pad_y)
    x0, x1 = max(0, x0 - pad_x), min(processed.shape[1], x1 + pad_x)

    cropped = Image.fromarray(processed).crop((x0, y0, x1, y1))
    cropped = cropped.resize((56, 56), Image.Resampling.LANCZOS)

    final = Image.new('L', (target_size, target_size), 255)
    final.paste(cropped, ((target_size-56)//2, (target_size-56)//2))
    final.save(output_path, "PNG")


# ──────────────────────────────────────────────────────────────
#  Update compare function to use new methods
# ──────────────────────────────────────────────────────────────
def compare_aggressive_methods(image_path):
    methods = [
        'mild_clahe_otsu',       # your current best "safe" one
        'strong_clahe_otsu',
        'niblack_tuned',         # tune k/window until bg stays mostly >0.8
        'background_subtract_otsu',
        'strong_clahe_niblack',
    ]
    # ... rest same as your compare_aggressive_methods ...
    # just change method names & call aggressive_preprocessing_v2(..., method=method, debug=False)
    
    fig, axes = plt.subplots(len(methods) + 1, 3, figsize=(12, 4*(len(methods)+1)))
    fig.suptitle(f'Aggressive Preprocessing Comparison\n{os.path.basename(image_path)}', 
                 fontsize=14)
    
    # Original
    img_orig = Image.open(image_path).convert('L')
    transform = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    orig_t = transform(img_orig)
    orig_np = (orig_t[0].numpy() * 0.5) + 0.5
    
    axes[0, 0].imshow(img_orig, cmap='gray')
    axes[0, 0].set_title('Original')
    axes[0, 0].axis('off')
    
    axes[0, 1].imshow(orig_np, cmap='gray', vmin=0, vmax=1)
    axes[0, 1].set_title(f'Original (Model Sees)\nMean={orig_np.mean():.3f}')
    axes[0, 1].axis('off')
    
    axes[0, 2].hist(orig_np.flatten(), bins=50, alpha=0.7, color='blue')
    axes[0, 2].set_title(f'Signature: {(orig_np<0.9).sum()/(64*64)*100:.1f}%')
    axes[0, 2].axvline(0.5, color='red', linestyle='--')
    axes[0, 2].set_xlim(0, 1)
    
    # Each method
    for idx, method in enumerate(methods, start=1):
        # Process
        import tempfile
        with tempfile.NamedTemporaryFile(suffix='.png', delete=False) as tmp:
            tmp_path = tmp.name
        
        aggressive_preprocessing(image_path, tmp_path, method=method)
        
        # Load result
        img_processed = Image.open(tmp_path)
        processed_t = transform(img_processed)
        processed_np = (processed_t[0].numpy() * 0.5) + 0.5
        
        # Visualize
        axes[idx, 0].imshow(img_processed, cmap='gray')
        axes[idx, 0].set_title(method.replace('_', ' ').title())
        axes[idx, 0].axis('off')
        
        axes[idx, 1].imshow(processed_np, cmap='gray', vmin=0, vmax=1)
        axes[idx, 1].set_title(f'Model Sees\nMean={processed_np.mean():.3f}')
        axes[idx, 1].axis('off')
        
        axes[idx, 2].hist(processed_np.flatten(), bins=50, alpha=0.7, color='green')
        axes[idx, 2].set_title(f'Signature: {(processed_np<0.9).sum()/(64*64)*100:.1f}%\nDark: {(processed_np<0.5).sum()/(64*64)*100:.1f}%')
        axes[idx, 2].axvline(0.5, color='red', linestyle='--')
        axes[idx, 2].set_xlim(0, 1)
        
        # Cleanup
        os.unlink(tmp_path)
        
        print(f"\n{method}:")
        print(f"  Mean: {processed_np.mean():.3f} (lower is better)")
        print(f"  Signature pixels (<0.9): {(processed_np<0.9).sum()/(64*64)*100:.1f}%")
        print(f"  Dark pixels (<0.5): {(processed_np<0.5).sum()/(64*64)*100:.1f}%")
    
    plt.tight_layout()
    plt.show()
    
    print("\n" + "="*60)
    print("RECOMMENDATION:")
    print("="*60)
    print("Choose the method where:")
    print("  1. Mean is LOWEST (more dark pixels)")
    print("  2. Signature is clearly visible")
    print("  3. No excessive noise")
    print("\nTypically 'adaptive_threshold' or 'clahe' work best")
    print("="*60)


def batch_aggressive_preprocess(input_dir, output_dir, method='adaptive_threshold'):
    """Batch process with chosen method"""
    
    import shutil
    
    print(f"\nBatch processing with method: {method}")
    print(f"Input:  {input_dir}")
    print(f"Output: {output_dir}")
    
    if os.path.exists(output_dir):
        response = input(f"\n{output_dir} exists. Delete? (y/n): ")
        if response.lower() == 'y':
            shutil.rmtree(output_dir)
        else:
            return
    
    os.makedirs(output_dir, exist_ok=True)
    
    files = [f for f in os.listdir(input_dir) 
             if f.endswith(('.png', '.PNG', '.jpg', '.jpeg'))]
    
    print(f"Processing {len(files)} files...")
    
    for idx, filename in enumerate(files):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)
        
        aggressive_preprocessing(input_path, output_path, method=method)
        
        if (idx + 1) % 100 == 0:
            print(f"  {idx + 1}/{len(files)}...")
    
    print(f"✓ Done! Processed {len(files)} files")


if __name__ == "__main__":
    print("""
    ╔════════════════════════════════════════════════════════════╗
    ║      AGGRESSIVE PREPROCESSING FOR FAINT SIGNATURES         ║
    ╚════════════════════════════════════════════════════════════╝
    
    Your signatures are very faint (Mean=0.968). Regular preprocessing
    doesn't help because the ORIGINAL scans are already light gray.
    
    This script applies aggressive thresholding to create darker,
    more visible signatures.
    
    USAGE:
    
    1. Test different methods on one image:
       compare_aggressive_methods('./content/PNG/Offline Forgeries/forgeries_10_13.png')
    
    2. After choosing best method, process all:
       batch_aggressive_preprocess(
           './content/PNG/Offline Genuine',
           './content/PNG_AGGRESSIVE/Offline Genuine',
           method='adaptive_threshold'  # or 'otsu', 'clahe', 'contrast_extreme'
       )
    """)
```

```python
"""
CRITICAL PREPROCESSING ISSUE DIAGNOSIS
======================================

Your stats show:
- Mean: 0.967 (97% white)
- Center: 0.773 (signature area is light gray, not dark)
- Background: 1.000 (pure white)

This means signatures are TOO FAINT for the model to learn from!
"""

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
import cv2

def diagnose_single_image(image_path):
    """Detailed diagnosis of one image"""
    
    print(f"\n{'='*60}")
    print(f"Diagnosing: {os.path.basename(image_path)}")
    print(f"{'='*60}")
    
    # Load original
    img_original = Image.open(image_path).convert('L')
    arr_original = np.array(img_original)
    
    print(f"\n1. ORIGINAL IMAGE:")
    print(f"   Size: {img_original.size}")
    print(f"   Min: {arr_original.min()}, Max: {arr_original.max()}")
    print(f"   Mean: {arr_original.mean():.1f}")
    print(f"   Median: {np.median(arr_original):.1f}")
    
    # Check histogram
    hist, bins = np.histogram(arr_original.flatten(), bins=256, range=(0, 256))
    
    print(f"\n2. PIXEL DISTRIBUTION:")
    print(f"   Pixels > 250 (white): {(arr_original > 250).sum() / arr_original.size * 100:.1f}%")
    print(f"   Pixels < 50 (black): {(arr_original < 50).sum() / arr_original.size * 100:.1f}%")
    print(f"   Pixels 50-250 (gray): {((arr_original >= 50) & (arr_original <= 250)).sum() / arr_original.size * 100:.1f}%")
    
    # Apply your current preprocessing
    from torchvision import transforms
    transform = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    
    img_processed = transform(img_original)
    img_processed_np = (img_processed[0].numpy() * 0.5) + 0.5  # Denormalize
    
    print(f"\n3. AFTER YOUR PREPROCESSING:")
    print(f"   Min: {img_processed_np.min():.3f}, Max: {img_processed_np.max():.3f}")
    print(f"   Mean: {img_processed_np.mean():.3f}")
    print(f"   Signature visibility: {(img_processed_np < 0.9).sum() / img_processed_np.size * 100:.1f}%")
    
    # Visualize
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    
    # Row 1: Original analysis
    axes[0, 0].imshow(arr_original, cmap='gray', vmin=0, vmax=255)
    axes[0, 0].set_title(f'Original\nMean={arr_original.mean():.0f}')
    axes[0, 0].axis('off')
    
    axes[0, 1].hist(arr_original.flatten(), bins=50, color='blue', alpha=0.7)
    axes[0, 1].set_title('Original Histogram')
    axes[0, 1].set_xlabel('Pixel Value')
    axes[0, 1].axvline(127, color='red', linestyle='--', label='Mid-gray')
    axes[0, 1].legend()
    
    # Otsu threshold to see what should be kept
    _, binary = cv2.threshold(arr_original, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    axes[0, 2].imshow(binary, cmap='gray')
    axes[0, 2].set_title('Otsu Threshold\n(What model should see)')
    axes[0, 2].axis('off')
    
    # Row 2: Processed analysis
    axes[1, 0].imshow(img_processed_np, cmap='gray', vmin=0, vmax=1)
    axes[1, 0].set_title(f'After Resize+Norm\nMean={img_processed_np.mean():.3f}')
    axes[1, 0].axis('off')
    
    axes[1, 1].hist(img_processed_np.flatten(), bins=50, color='green', alpha=0.7)
    axes[1, 1].set_title('Processed Histogram')
    axes[1, 1].set_xlabel('Pixel Value (0-1)')
    axes[1, 1].axvline(0.5, color='red', linestyle='--', label='Mid-gray')
    axes[1, 1].legend()
    
    # Show what the model "sees"
    # Anything > 0.9 is basically invisible to the model
    signature_mask = img_processed_np < 0.9
    axes[1, 2].imshow(signature_mask, cmap='gray')
    axes[1, 2].set_title(f'What Model Sees\n{signature_mask.sum()/(64*64)*100:.1f}% of pixels')
    axes[1, 2].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # DIAGNOSIS
    print(f"\n{'='*60}")
    print("DIAGNOSIS:")
    print(f"{'='*60}")
    
    if img_processed_np.mean() > 0.95:
        print("❌ PROBLEM: Image is too washed out (>95% white)")
        print("   The signature is barely visible to the model!")
        
    if (img_processed_np < 0.9).sum() / img_processed_np.size < 0.15:
        print("❌ PROBLEM: Signature covers <15% of image")
        print("   Most of the image is useless white space")
        
    if img_processed_np.min() > 0.5:
        print("❌ PROBLEM: No dark pixels (min > 0.5)")
        print("   Signature strokes aren't dark enough")
        
    print(f"\n{'='*60}")
    print("RECOMMENDATIONS:")
    print(f"{'='*60}")
    print("Your preprocessing needs to:")
    print("1. INCREASE CONTRAST - make signature darker")
    print("2. BETTER CROPPING - remove excess white space")
    print("3. OPTIONAL: Apply adaptive thresholding")
    print(f"{'='*60}")


def check_multiple_images(directory, num_samples=5):
    """Check multiple images"""
    
    files = [f for f in os.listdir(directory) if f.endswith(('.png', '.PNG', '.jpg', '.jpeg'))]
    
    print(f"\nChecking {num_samples} images from {directory}...")
    
    for i, fname in enumerate(files[:num_samples]):
        diagnose_single_image(os.path.join(directory, fname))
        
        if i < num_samples - 1:
            input("Press Enter to continue to next image...")


def compare_preprocessing_methods(image_path):
    """Compare different preprocessing approaches"""
    
    img = Image.open(image_path).convert('L')
    arr = np.array(img)
    
    # Method 1: Current (just resize)
    from torchvision import transforms
    transform_current = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    current = transform_current(img)
    current_np = (current[0].numpy() * 0.5) + 0.5
    
    # Method 2: With contrast enhancement
    from PIL import ImageEnhance
    enhancer = ImageEnhance.Contrast(img)
    img_enhanced = enhancer.enhance(1.5)  # 50% more contrast
    
    transform_enhanced = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    enhanced = transform_enhanced(img_enhanced)
    enhanced_np = (enhanced[0].numpy() * 0.5) + 0.5
    
    # Method 3: With Otsu threshold
    _, binary = cv2.threshold(arr, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    binary_inverted = 255 - binary  # Back to white background
    img_binary = Image.fromarray(binary_inverted)
    
    transform_binary = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    binarized = transform_binary(img_binary)
    binarized_np = (binarized[0].numpy() * 0.5) + 0.5
    
    # Method 4: Proper crop + enhance
    # Find signature bounds
    coords = np.where(binary > 10)
    if len(coords[0]) > 0:
        y0, y1 = coords[0].min(), coords[0].max() + 1
        x0, x1 = coords[1].min(), coords[1].max() + 1
        
        # Add 15% padding
        h, w = y1 - y0, x1 - x0
        pad_y, pad_x = int(h * 0.15), int(w * 0.15)
        y0 = max(0, y0 - pad_y)
        y1 = min(arr.shape[0], y1 + pad_y)
        x0 = max(0, x0 - pad_x)
        x1 = min(arr.shape[1], x1 + pad_x)
        
        cropped = Image.fromarray(arr).crop((x0, y0, x1, y1))
        
        # Enhance contrast
        enhancer = ImageEnhance.Contrast(cropped)
        cropped = enhancer.enhance(1.3)
        
        # Resize to 56x56, then pad to 64x64
        cropped = cropped.resize((56, 56), Image.Resampling.LANCZOS)
        final = Image.new('L', (64, 64), 255)
        final.paste(cropped, (4, 4))
        
        proper = transform_current(final)
        proper_np = (proper[0].numpy() * 0.5) + 0.5
    else:
        proper_np = current_np
    
    # Visualize comparison
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    fig.suptitle('Preprocessing Method Comparison', fontsize=16)
    
    methods = [
        ('Current (Just Resize)', current_np),
        ('+ Contrast Enhance', enhanced_np),
        ('+ Otsu Threshold', binarized_np),
        ('Crop + Enhance (Best)', proper_np)
    ]
    
    for idx, (name, img_np) in enumerate(methods):
        # Show image
        axes[0, idx].imshow(img_np, cmap='gray', vmin=0, vmax=1)
        axes[0, idx].set_title(f'{name}\nMean={img_np.mean():.3f}')
        axes[0, idx].axis('off')
        
        # Show histogram
        axes[1, idx].hist(img_np.flatten(), bins=50, alpha=0.7)
        axes[1, idx].set_xlim(0, 1)
        axes[1, idx].axvline(0.5, color='red', linestyle='--', alpha=0.5)
        axes[1, idx].set_title(f'Signature: {(img_np < 0.9).sum()/(64*64)*100:.1f}%')
    
    plt.tight_layout()
    plt.show()
    
    print("\nSTATISTICS COMPARISON:")
    print("="*60)
    for name, img_np in methods:
        print(f"\n{name}:")
        print(f"  Mean: {img_np.mean():.3f}")
        print(f"  Min: {img_np.min():.3f}, Max: {img_np.max():.3f}")
        print(f"  Signature pixels: {(img_np < 0.9).sum()/(64*64)*100:.1f}%")
        print(f"  Dark pixels (<0.5): {(img_np < 0.5).sum()/(64*64)*100:.1f}%")


if __name__ == "__main__":
    print("""
    ╔════════════════════════════════════════════════════════════╗
    ║  PREPROCESSING DIAGNOSIS TOOLKIT                          ║
    ║  Use this to identify why your model only gets 60% acc   ║
    ╚════════════════════════════════════════════════════════════╝
    """)
    
    print("\nUsage:")
    print("1. Diagnose single image:")
    print("   diagnose_single_image('./content/PNG/Offline Genuine/001_01.png')")
    print("\n2. Check multiple images:")
    print("   check_multiple_images('./content/PNG/Offline Genuine', num_samples=5)")
    print("\n3. Compare preprocessing methods:")
    print("   compare_preprocessing_methods('./content/PNG/Offline Genuine/001_01.png')")
```
