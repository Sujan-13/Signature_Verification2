```python
import shutil
import cv2
import os
from PIL import Image
import numpy as np
import torchvision.transforms as transforms
def crop_and_save_signature(input_path, output_path, label):
    size = 96
    target_size = 80
    img = Image.open(input_path).convert('L')

    arr = np.array(img)

    # ← Use Otsu only for FINDING coordinates, not for the output image
    _, binary_arr = cv2.threshold(arr, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    # binary_arr is now: signature=255 (white), background=0 (black)

    coords = np.where(binary_arr > 10)  # find signature pixels
    if len(coords[0]) == 0:
        img.resize((size, size), Image.Resampling.LANCZOS).save(output_path)
        return

    y0, y1 = coords[0].min()+2, coords[0].max() + 3 #margi
    x0, x1 = coords[1].min()+2, coords[1].max() + 3

    h, w = y1 - y0, x1 - x0
    pad_y = int(h * 0.1)
    pad_x = int(w * 0.1)
    y0 = max(0, y0 - pad_y)
    y1 = min(arr.shape[0], y1 + pad_y)
    x0 = max(0, x0 - pad_x)
    x1 = min(arr.shape[1], x1 + pad_x)

    # ← Crop from ORIGINAL arr, not binary_arr
    cropped = Image.fromarray(arr).crop((x0, y0, x1, y1))

    # ← Re-binarize the cropped region with a fixed threshold
    cropped_arr = np.array(cropped)
    cropped_arr[cropped_arr > 220] = 255  # background → pure white
    cropped_arr[cropped_arr <= 220] = 0   # signature → pure black
    cropped = Image.fromarray(cropped_arr)

    cropped = cropped.resize((target_size, target_size), Image.Resampling.LANCZOS)

    bg = Image.new('L', (size, size), 255)
    x = (size - target_size) // 2
    y = (size - target_size) // 2
    bg.paste(cropped, (x, y))
    # if label:
    #     augment = transforms.Compose([
    #         transforms.Resize((64, 64)),
    #         # transforms.RandomAffine(degrees=10, translate=(0.05, 0.05), scale=(0.9, 1.1), fill=255),
    #         # transforms.RandomPerspective(distortion_scale=0.3, p=0.3, fill=255),
    #         transforms.RandomAffine(degrees=5, translate=(0.03, 0.03), scale=(0.95, 1.05), fill=255),
    #         transforms.RandomPerspective(distortion_scale=0.15, p=0.3, fill=255),
    #         transforms.GaussianBlur(3, sigma=(0.1, 1.0)), 
    #     ])
    #     bg = augment(bg)
    bg.save(output_path, "PNG", optimize=True) 



# Run on your dataset
def pngfy(directories):
    for input_dir, output_dir,label in directories:
        
        if os.path.exists(output_dir):
            shutil.rmtree(output_dir)   # ← nukes everything
            print(f"Deleted old folder: {output_dir}" )
        os.makedirs(output_dir, exist_ok=True)
        
        for file in os.listdir(input_dir):
                if file.endswith(('.PNG')):
                    crop_and_save_signature(os.path.join(input_dir, file),
                                        os.path.join(output_dir, file.split('.')[0] + ".PNG"),label)
                if file.endswith(('.png', '.jpg', '.jpeg', '.bmp')):
                     crop_and_save_signature(os.path.join(input_dir, file),
                                        os.path.join(output_dir, file.split('.')[0] + ".png"),label)

    
    print("All datasets processed perfectly!")

```
