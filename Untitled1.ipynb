```python

import os
import re

def rename_signatures(folder_path):
    # Regex to match filenames like "forgeries_0A_2.png" or "forgeries_01_3.jpg"
    # Assumes author ID is 2 characters (hex-like or decimal), sig number is digits, extension is image
    # pattern = re.compile(r'^forgeries_([0-9A-Fa-f]{2})_(\d+)\.(\w+)$')
    pattern = re.compile(r'^original_([0-9A-Fa-f]{1,2})_(\d+)\.(\w+)$')

    for filename in os.listdir(folder_path):
        match = pattern.match(filename)
        if match:
            author_str, sig_num, ext = match.groups()
            
            # Parse author as hex to int (e.g., "0A" -> 10, "01" -> 1)
            try:
                author_num = int(author_str, 16)  # Base 16 for hex
            except ValueError:
                print(f"Skipping invalid author ID: {author_str} in {filename}")
                continue
            
            # Add 20
            new_author_num = author_num + 20
            
            # Convert back to string, pad to 3 digits with leading zeros (decimal)
            new_author_str = f"{new_author_num:03d}"
            
            # New filename: e.g., "030_2.png"
            new_filename = f"{new_author_str}_{sig_num}.{ext}"
            
            # Rename
            old_path = os.path.join(folder_path, filename)
            new_path = os.path.join(folder_path, new_filename)
            os.rename(old_path, new_path)
            print(f"Renamed {filename} to {new_filename}")
        else:
            print(f"Skipping non-matching file: {filename}")

# Example usage: replace with your folder path
rename_signatures(r"C:\Users\gupta\Downloads\archive\signatures\full_org")

```
