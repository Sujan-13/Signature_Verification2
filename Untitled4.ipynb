```python
import torch
import pprint
import re

path = r"C:\Users\gupta\OneDrive\Desktop\best_model.pt"
print(f"🔍 Inspecting {path}\n")

checkpoint = torch.load(path, weights_only=False, map_location="cpu")

for key, value in checkpoint.items():
    print(f"📌 Key: {key}")
    
    if isinstance(value, (int, float, str, bool)):
        print(f"   → Value: {value}")
    
    elif isinstance(value, dict):
        print(f"   → Dict with {len(value)} items")
        if len(value) < 20:  # small dicts are safe to show
            pprint.pprint(value, indent=4)
        else:
            print(f"   → First 5 keys: {list(value.keys())[:5]}")
            
        # Special check: scan for Git commit hash (7-40 hex chars)
        for k, v in value.items():
            if isinstance(v, str) and re.search(r'\b[0-9a-f]{7,40}\b', v):
                print(f"   🎉 POSSIBLE GIT HASH FOUND inside '{k}': {v}")
    
    elif isinstance(value, torch.Tensor):
        print(f"   → Tensor | shape: {value.shape} | dtype: {value.dtype}")
    
    elif isinstance(value, (list, tuple)):
        print(f"   → List/Tuple of length {len(value)}")
        if len(value) < 10:
            print(f"   → Content: {value}")
    
    else:
        print(f"   → Type: {type(value)} (size unknown)")
    
    # Global hash search in any string representation
    if isinstance(value, str) and re.search(r'\b[0-9a-f]{7,40}\b', value):
        print(f"   🎉 POSSIBLE GIT HASH FOUND: {value}")
    
    print("-" * 60)

print("\n✅ Done! Look for any 🎉 line above — that would be your commit hash.")
```
